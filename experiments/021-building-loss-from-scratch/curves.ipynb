# Imports

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd
```

# Params

```python
EXP_DIR = Path(".").resolve()
ARTIFACTS_DIR = EXP_DIR / "artifacts"
FIGURES_DIR = EXP_DIR / "figures"

ARCHITECTURES = ["unet-resnet18", "unet-resnet34", "lraspp-mobilenetv3"]
LOSSES = ["current", "building", "structure"]
SEED = 42
STEPS = 2500
LOSS_COLOR = {"current": "tab:blue", "building": "tab:orange", "structure": "tab:green"}

# column suffix -> axis label; order is the order plots are produced in
METRICS = [
    ("miou", "mIoU"),
    ("loss", "loss"),
    ("mdice", "mean Dice"),
    ("building_missed_share", "buildings missed"),
    ("track_cldice", "road connectivity, clDice"),
    ("track_fragments_per_km", "road fragments per km"),
    ("track_skeleton_recall", "road pixels found, skeleton recall"),
]
```

# Data

```python
def load_history(architecture: str, loss: str, seed: int, steps: int) -> pd.DataFrame:
    path = ARTIFACTS_DIR / f"history-{architecture}-{loss}-seed{seed}-steps{steps}.csv"
    return pd.read_csv(path)
```

```python
histories = {
    (architecture, loss): load_history(architecture, loss, SEED, STEPS)
    for architecture in ARCHITECTURES
    for loss in LOSSES
}
```

# Plots

```python
def plot_model_curves(ax: plt.Axes, architecture: str, metric: str) -> None:
    for loss in LOSSES:
        history = histories[(architecture, loss)]
        color = LOSS_COLOR[loss]
        ax.plot(history["step"], history[f"train_{metric}"], color=color, linestyle="--", alpha=0.6)
        ax.plot(history["step"], history[f"val_{metric}"], color=color, linestyle="-")
```

```python
def plot_metric_figure(architecture: str, metric: str, label: str) -> None:
    fig, ax = plt.subplots(figsize=(6, 4))
    plot_model_curves(ax, architecture, metric)
    ax.set_title(f"{architecture} — {label}")
    ax.set_xlabel("step")
    ax.set_ylabel(label)
    loss_handles = [plt.Line2D([], [], color=LOSS_COLOR[loss], label=loss) for loss in LOSSES]
    style_handles = [plt.Line2D([], [], color="black", linestyle="-", label="val"),
                      plt.Line2D([], [], color="black", linestyle="--", alpha=0.6, label="train")]
    ax.legend(handles=loss_handles + style_handles, fontsize=8)
    fig.tight_layout()
    fig.savefig(FIGURES_DIR / f"train-val-{metric}-{architecture}-seed{SEED}.png", dpi=150, bbox_inches="tight")
    plt.show()
```
